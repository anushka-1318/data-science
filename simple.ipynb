**factorial**

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

# Example usage:
number = 5
result = factorial(number)
print(f"The factorial of {number} is {result}")

number = 0
result = factorial(number)
print(f"The factorial of {number} is {result}")

number = 7
result = factorial(number)
print(f"The factorial of {number} is {result}")
```

table

```python
def display_multiplication_table(number):
    print(f"Multiplication Table for {number}:")
    for i in range(1, 11): # Generates table from 1 to 10
        print(f"{number} x {i} = {number * i}")

# Example usage:
num_to_display = 9
display_multiplication_table(num_to_display)
```

odd even number

```python
def check_odd_even(number):
    if number % 2 == 0:
        print(f"The number {number} is Even.")
    else:
        print(f"The number {number} is Odd.")

# Example usage:
num_to_check = 10
check_odd_even(num_to_check)

num_to_check = 7
check_odd_even(num_to_check)

num_to_check = 0
check_odd_even(num_to_check)
```

palindrome

```python
def is_palindrome(number):
    # Convert the number to a string to easily reverse it
    num_str = str(number)
    # Check if the string is equal to its reverse
    if num_str == num_str[::-1]:
        print(f"The number {number} is a palindrome.")
    else:
        print(f"The number {number} is not a palindrome.")

# Example usage:
num_to_check_palindrome = 121
is_palindrome(num_to_check_palindrome)

num_to_check_palindrome = 12345
is_palindrome(num_to_check_palindrome)

num_to_check_palindrome = 1221
is_palindrome(num_to_check_palindrome)
```

## Define Problem Domain

### Subtask:
First, let's define a simple problem domain for the expert system. For example, it could recommend an activity based on weather conditions, diagnose a simple issue, or recommend a type of drink. We'll start by outlining the structure to define facts and rules.


### Chosen Problem Domain: Drink Recommendation System

This expert system will recommend a drink based on the user's current mood and the prevailing weather conditions.

#### Key Facts (Inputs):
*   **Mood**: Represents the user's current emotional state. Possible values could be: `happy`, `tired`, `stressed`.
*   **Weather**: Describes the current weather conditions. Possible values could be: `hot`, `cold`, `mild`.

#### Outline of Rules (If-Then Statements):
*   **Rule 1**: IF `Mood` is `happy` AND `Weather` is `hot` THEN Recommend: "Iced Tea"
*   **Rule 2**: IF `Mood` is `tired` AND `Weather` is `cold` THEN Recommend: "Hot Coffee"
*   **Rule 3**: IF `Mood` is `stressed` AND `Weather` is `mild` THEN Recommend: "Herbal Tea"
*   **Rule 4**: IF `Mood` is `happy` AND `Weather` is `cold` THEN Recommend: "Hot Chocolate"
*   **Rule 5**: IF `Mood` is `tired` AND `Weather` is `hot` THEN Recommend: "Cold Brew"
*   **Rule 6**: IF `Mood` is `stressed` AND `Weather` is `hot` THEN Recommend: "Lemonade"

## Develop Knowledge Base

### Subtask:
Implement a structure to store the 'knowledge' of the expert system, specifically the rules defined in the previous step, using a suitable Python data structure.


**Reasoning**:
The subtask requires defining a Python variable to store the rules, representing each rule as a dictionary, and then storing all defined rules as a list of dictionaries in the knowledge_base variable. This code block will achieve that by creating a list of dictionaries.



```python
knowledge_base = [
    {'mood': 'happy', 'weather': 'hot', 'recommendation': 'Iced Tea'},
    {'mood': 'stressed', 'weather': 'cold', 'recommendation': 'Hot Chocolate'},
    {'mood': 'energetic', 'weather': 'mild', 'recommendation': 'Coffee'},
    {'mood': 'relaxed', 'weather': 'any', 'recommendation': 'Herbal Tea'},
    {'mood': 'adventurous', 'weather': 'any', 'recommendation': 'Smoothie'},
    {'mood': 'celebratory', 'weather': 'any', 'recommendation': 'Sparkling Water'}
]

print("Knowledge Base created successfully:")
for rule in knowledge_base:
    print(rule)
```

```python
knowledge_base = [
    {'mood': 'happy', 'weather': 'hot', 'recommendation': 'Iced Tea'},
    {'mood': 'tired', 'weather': 'cold', 'recommendation': 'Hot Coffee'},
    {'mood': 'stressed', 'weather': 'mild', 'recommendation': 'Herbal Tea'},
    {'mood': 'happy', 'weather': 'cold', 'recommendation': 'Hot Chocolate'},
    {'mood': 'tired', 'weather': 'hot', 'recommendation': 'Cold Brew'},
    {'mood': 'stressed', 'weather': 'hot', 'recommendation': 'Lemonade'}
]

print("Updated Knowledge Base created successfully:")
for rule in knowledge_base:
    print(rule)
```

## Build Inference Mechanism

### Subtask:
Develop a function that takes user input (facts) and applies the rules from the knowledge base to infer a decision. For a simple system, this will involve checking conditions and returning the corresponding decision.


**Reasoning**:
The subtask requires developing an inference mechanism by creating a Python function that uses the previously defined `knowledge_base` to provide drink recommendations based on user mood and weather. The function will iterate through the rules, match user inputs, and return a recommendation or a default message. After defining the function, it will be tested with example inputs.



```python
def infer_drink_recommendation(user_mood, user_weather):
    for rule in knowledge_base:
        if rule['mood'] == user_mood and rule['weather'] == user_weather:
            return rule['recommendation']
    return 'No specific recommendation based on the current knowledge base. Try something common like Water or Juice.'

# Test the function with example inputs
print(f"Recommendation for happy and hot weather: {infer_drink_recommendation('happy', 'hot')}")
print(f"Recommendation for tired and cold weather: {infer_drink_recommendation('tired', 'cold')}")
print(f"Recommendation for stressed and hot weather: {infer_drink_recommendation('stressed', 'hot')}")
print(f"Recommendation for angry and rainy weather: {infer_drink_recommendation('angry', 'rainy')}") # Test for no match
```

## Design User Interface

### Subtask:
Create a simple interface for the user to provide input (e.g., answers to questions about the problem domain) and for the system to display the resulting decision.


### Implement User Interface

**Reasoning**:
This step will create a user-friendly interface that prompts the user for their mood and weather conditions. It will then use the `infer_drink_recommendation` function developed earlier to provide a drink recommendation based on the user's input. The input will be converted to lowercase to ensure case-insensitive matching with the knowledge base.

**Reasoning**:
Implement the user interface by prompting the user for mood and weather, then use the `infer_drink_recommendation` function to provide a recommendation.



```python
user_mood_input = input("Enter your current mood (happy, tired, stressed): ").lower()
user_weather_input = input("Enter current weather conditions (hot, cold, mild): ").lower()

recommendation = infer_drink_recommendation(user_mood_input, user_weather_input)

print(f"\nBased on your mood ('{user_mood_input}') and the weather ('{user_weather_input}'), I recommend: {recommendation}")
```
